```python
class Car :
  def __init__(self):
    self.clutch = False
    self.acc = False
    self.brk = False

  def start(self):
    self.clutch = True
    self.acc = True
    self.brk = False
    print("the Car is started")

  def stop(self):
    self.clutch = False
    self.acc = False
    self.brk = True
    print("the Car is stopped")


alto = Car()
alto.start()
alto.stop()
```

```python
from abc import ABC, abstractmethod

class beautysalon(ABC):
  @abstractmethod
  def perform(self):
    pass

class makeup_service(beautysalon):
  def perform(self):
    print("makeup service")

class massage_service(beautysalon):
  def perform(self):
    print("massage service")

class facial_service(beautysalon):
  def perform(self):
    print("facial service")

makeup = makeup_service()
massage = massage_service()
facial = facial_service()

makeup.perform()
massage.perform()
facial.perform()
```

```python
import collections
#task 14 restaurant ordering system

menu = {
          "burger" : 500,
          "pizza" : 1200,
          "pasta" : 800,
          "drink" : 200
    }

print("========== RESTAURANT MENU ==========")
for key,value in menu.items():
      print(f"{key} : {value}")

orders = [] # Initialize orders list

while True :
    item_name = input("enter item name : ").title()
    if item_name in menu:
        quantity = int(input("enter quantity of your item; "))
        total_price = menu[item_name] * quantity
        print(f"total price for {item_name}: {total_price}")

        orders.append({
             "item": item_name,
             "quantity" : quantity,
             "price_per_item": menu[item_name], # Added price per item for better receipt
             "total_price" : total_price
        })
    else:
        print("item not found in menu")

    choice = input("do you want to add more items? (yes/no) : ")
    if choice.lower() == "no":
        break

calculated_total = 0 # Use a different variable name to avoid confusion with 'total' after discount
for order in orders:
    calculated_total += order["total_price"]

tax = calculated_total * 13 / 100

discount = 0
if calculated_total >= 3000:
  discount = calculated_total * 10 / 100

total_after_discount = calculated_total - discount
grand_total = total_after_discount + tax

def receipt():
      print("========== RECEIPT ==========")

      for order in orders:
        # Corrected f-string to use existing keys for clarity
        print(f"{order['item']} : {order['quantity']} x {order['price_per_item']} = {order['total_price']}")

      print(f"Subtotal: {calculated_total:.2f}")
      print(f"Tax (13%): {tax:.2f}")
      print(f"Discount (10% on orders >= 3000): {discount:.2f}")
      print(f"Total (after discount and tax): {grand_total:.2f}")
receipt()
```
