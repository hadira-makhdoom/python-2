```python
#restaurant ordering system

menu = {
    "Burger": 500,
    "pizza": 1200,
    "pasta": 800,
    "drink": 200
}

orders = []

def display_receipt(orders):
  subtotal = 0

  print("\n============== RECEIPT ================")


  for order in orders:
    item = order["item"]
    quantity = order["quantity"]
    price = order["price"]
    item_total = order["subtotal"]

    print(f"{item} x {quantity} = Rs.{item_total}")
    subtotal += item_total

  tax = subtotal * 13 / 100
  total = subtotal + tax

  discount = 0
  if total >=3000:
    discount = total * 0.10

  final_bill = total - discount

  print("_________________________")
  print(f"subtotal     : Rs. {subtotal:.2f}")
  print(f"tax (13%)    : Rs. {tax:.2f}")
  print(f"discount (10%) : Rs. {discount:.2f}")
  print(f"total        : Rs. {final_bill:.2f}")
  print("============================")


  #display menu
  print("============= RESTAURANT MENU ================")

  for item, price in menu.items():
    print(f"{item} : Rs. {price}")

  #take multiple orders
  while True:
    item = input("\nEnter item name (or 'done' to finish): ")

    if item.lower() == "done":
      break

    if item in menu:
      quantity = int(input("Enter quantity: "))
      price = menu[item]
      subtotal = price * quantity
      orders.append({"item": item, "quantity": quantity, "price": price, "subtotal": subtotal})
    else:
      print("Invalid item. Please try again.")

  #generate receipt
  if len(orders) > 0:
    display_receipt(orders)
  else:
    print("No orders placed.")
```
